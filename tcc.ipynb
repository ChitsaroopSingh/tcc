Importing libraries

```python
import pandas as pd
import numpy as np
```

Importing the dataset

```python
tcc = pd.read_csv("dataset.csv")
tcc.info()
```

Data Cleaning

Removing useless labels

```python
tcc=tcc.drop(['CustomerID','Count','Country','State','City','Zip Code','Lat Long','Latitude','Longitude','Churn Label','Churn Score','CLTV','Churn Reason'],axis=1)
tcc.info()
```

Converting string datatype to numeric

```python
tcc['Total Charges']=pd.to_numeric(tcc['Total Charges'], errors='coerce')
tcc['Total Charges'].isna().sum()  
tcc['Total Charges'] = tcc['Total Charges'].fillna(0)
tcc.info()
```

Encoding categorical labels

```python
binary_cols = ['Gender', 'Senior Citizen', 'Partner', 'Dependents', 'Phone Service', 'Paperless Billing']
for col in binary_cols:
    tcc[col]=tcc[col].map({'Yes':1,'No':0,
                           'Male':1,'Female':0})
tcc.info()
```

Dropping duplicate values

```python
tcc = tcc.drop_duplicates()
tcc.info()
```

Encoding remaining categorical columns

```python
categorical_cols = tcc.select_dtypes(include="object").columns

tcc = pd.get_dummies(tcc, columns=categorical_cols, drop_first=True, dtype=int)
tcc.info()
```

Separating Churn Value from dataset as it is the label we have to predict

```python
X = tcc.drop("Churn Value",axis=1)
y=tcc["Churn Value"]
print(X.shape)
print(y.shape)
```
