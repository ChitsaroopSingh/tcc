Importing libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("dataset.csv")

df.head()
```

Shape of data

```python
print(df.shape)
print(df.info())
```

Churn Distribution

```python
churn_counts = df["Churn Value"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(["No Churn", "Churn"], churn_counts.values)

plt.title("Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

Churn vs Contract Type

```python
contract_churn = pd.crosstab(df["Contract"], df["Churn Value"])

x = range(len(contract_churn))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(
    [i - width/2 for i in x],
    contract_churn[0],
    width,
    label="No Churn"
)

plt.bar(
    [i + width/2 for i in x],
    contract_churn[1],
    width,
    label="Churn"
)

plt.xticks(x, contract_churn.index)
plt.title("Churn vs Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.legend()

plt.show()
```

Churn vs Monthly Charges

```python
plt.figure(figsize=(7, 5))

sns.boxplot(data=df, x="Churn Value", y="Monthly Charges")

plt.title("Churn vs Monthly Charges")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")

plt.show()
```

Churn vs Customer Tenure

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Churn Value", y="Tenure Months")

plt.title("Churn vs Customer Tenure")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.show()
```

Churn vs Internet Service

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Internet Service", hue="Churn Value")

plt.title("Churn vs Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")

plt.show()
```
