Importing libraries

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("dataset.csv")

df.head()
```

Shape of data

```python
print(df.shape)
print(df.info())
```

Churn Distribution

```python
churn_counts = df["Churn Value"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(["No Churn", "Churn"], churn_counts.values)

plt.title("Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```
